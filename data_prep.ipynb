## Cleaning

Dropping duplicates, keeping only confirmed incidents, and filtering VCDB to reasonable years. I'm not imputing the missing records-exposed values since the missingness isn't random -- I'll just flag it in the analysis.

```python
# Cleaning
vcdb = vcdb.drop_duplicates()
```

```python
vcdb = vcdb[vcdb['security_incident.Confirmed'] == True]
```

```python
vcdb = vcdb[vcdb['timeline.incident.year'].between(2000, 2025)]
```

```python
print("Rows with records exposed data:", vcdb['attribute.confidentiality.data_total'].notna().sum())
print("Rows without:", vcdb['attribute.confidentiality.data_total'].isna().sum())
```

## Multi-Value Fields

Some rows in CyberEvents have multiple motives or subtypes in a single comma-separated string. Exploding these into separate rows lets me count each value properly without undercounting.

```python
cyb_motive = cyb.copy()

cyb_motive['motive'] = cyb_motive['motive'].str.split(',')

cyb_motive = cyb_motive.explode('motive')

cyb_motive['motive'] = cyb_motive['motive'].str.strip()
```

```python
cyb_motive.info()
```

```python
cyb_subtype = cyb.copy()

cyb_subtype['event_subtype'] = cyb_subtype['event_subtype'].str.split(',')

cyb_subtype = cyb_subtype.explode('event_subtype')

cyb_subtype['event_subtype'] = cyb_subtype['event_subtype'].str.strip()
cyb_subtype
```
