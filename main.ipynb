```python
# Cleaning
vcdb = vcdb.drop_duplicates()
```

```python
vcdb = vcdb[vcdb['security_incident.Confirmed'] == True]
```

```python
vcdb = vcdb[vcdb['timeline.incident.year'].between(2000, 2025)]
```

```python
print("Rows with records exposed data:", vcdb['attribute.confidentiality.data_total'].notna().sum())
print("Rows without:", vcdb['attribute.confidentiality.data_total'].isna().sum())
```

```python
kev['dateAdded'] = pd.to_datetime(kev['dateAdded'])
```

```python
kev['year_added'] = kev['dateAdded'].dt.year
```

```python
print(kev['knownRansomwareCampaignUse'].value_counts())
```

```python
cyb = cyb.drop_duplicates(subset='slug')
```

```python
cyb['event_date'] = pd.to_datetime(cyb['event_date'])
```

```python
cyb['motive'] = cyb['motive'].str.strip().str.title()
print(cyb['motive'].value_counts())
```
